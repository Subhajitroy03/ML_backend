```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('himalayan_glof_data.csv')

sns.set_theme(style="whitegrid")

# 1. Distribution of the Risk Index
plt.figure(figsize=(10, 5))
sns.histplot(df['Risk_Index'], kde=True, bins=20, color='teal')
plt.title('Distribution of GLOF Risk Index')
plt.xlabel('Risk Index (0 to 1)')
plt.ylabel('Frequency')
plt.show()

# 2. Correlation Heatmap
plt.figure(figsize=(8, 6))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='RdBu_r', center=0)
plt.title('Feature Correlation Heatmap')
plt.show()

# 3. Pairplot to see relationships between features
sns.pairplot(df, vars=['Lake_Area_km2', 'Dam_Slope_deg', 'Lake_Temp_C', 'Elevation_m'], 
             hue='Risk_Index', palette='viridis')
plt.suptitle('Feature Relationships Colored by Risk', y=1.02)
plt.show()

# 4. Boxplots to check for outliers
plt.figure(figsize=(12, 6))
df_melted = df.drop('Risk_Index', axis=1).melt()
sns.boxplot(x='variable', y='value', data=df_melted)
plt.title('Range and Outliers for Input Parameters')
plt.xticks(rotation=45)
plt.show()
```
